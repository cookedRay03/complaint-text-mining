```python
# Menimpa df dengan data yang baru
df = pd.read_csv('./dataset/2_content_dataset_keluhan.csv')
```

```python
def clean_text(text):
    if pd.isna(text):
        return ""

    # Mengubah teks menjadi lowercase
    text = text.lower()

    # Menghapus URL
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # Menghapus angka
    text = re.sub(r'\d+', '', text)

    # Menghapus simbol, tanda baca, emoji, dan karakter non-huruf
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    # Menghapus spasi berlebih
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
df_cleaned = df.copy()

df_cleaned['content'] = df_cleaned['content'].apply(clean_text)

df_cleaned.head()
```

```python
df_cleaned.to_csv('./dataset/3_dataset_keluhan_cleaned.csv', index=False)
print("Dataset berhasil disimpan!")
print("Jumlah data:", len(df_cleaned))
```
